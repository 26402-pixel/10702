```python
def calculate(a, operator, b):
    if operator == '+':
        return a + b
    elif operator == '-':
        return a - b
    elif operator == '*':
        return a * b
    elif operator == '/':
        if b == 0:
            return "0으로 나눌 수 없습니다."
        return a / b
    else:
        return "지원하지 않는 연산자입니다."
print(calculate(10, '+', 5))
```
