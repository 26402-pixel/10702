```python
import datetime
from datetime import datetime,date,time
print('<디데이(d-day)프로그램')
print()
year=int(input('년도'))
month=int(input('월'))
day=int(input('일을 입력하세요:'))
dt4=date(year,month,day)
today=date.today()
dday=(dt4-today).days
print(f'{year}년{month}월{dday}일 남았습니다.')
```
