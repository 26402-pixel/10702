```python
import random
num=0
com=random.randint(1,10)
while num != com:
  num=int(input('1에서 10사이의 수를 입력하세요'))
  if num<com:
    print('%d보다 큽니다.'%num)
  elif num>com:
    print('%d보다 작습니다.'%num)
print('정답입니다.')
```
